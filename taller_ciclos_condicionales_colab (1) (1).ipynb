### Reto 01: Clasificador Tarifario con Descuentos Escalonados
**Contexto:** Una plataforma de transporte aplica tarifas según edad y horario pico. Si la edad es menor a 12 o mayor o igual a 65, se otorga 50% de descuento. Si es un horario pico (`es_pico == True`), se aplica un recargo del 20% sobre la tarifa base de $10.000 antes de cualquier descuento.

**Requerimiento:** Retorna el valor final a pagar (flotante redondeado a 2 decimales).

**Firma requerida:** `reto_01(edad: int, es_pico: bool)`

```python
def reto_01(edad: int, es_pico: bool):
    tarifa = 10000.0


    if es_pico:
        tarifa = tarifa * 1.20


    if edad < 12 or edad >= 65:
        tarifa = tarifa * 0.50

    return round(tarifa, 2)
```

### Reto 02: Validador Lógico de Triángulos y Clasificación
**Contexto:** Para que tres segmentos $a, b, c$ formen un triángulo, deben cumplir la desigualdad triangular: la suma de dos lados cualesquiera debe ser estrictamente mayor que el tercero ($a+b>c \land a+c>b \land b+c>a$).

**Requerimiento:** Si no forman triángulo retorna `'NO ES TRIANGULO'`. Si forman, clasifícalo en `'EQUILATERO'` (3 lados iguales), `'ISOSCELES'` (2 lados iguales) o `'ESCALENO'` (3 lados distintos).

**Firma requerida:** `reto_02(a: float, b: float, c: float)`

```python
def reto_02(a: float, b: float, c: float):

    if not (a > 0 and b > 0 and c > 0 and (a + b > c) and (a + c > b) and (b + c > a)):
        return 'NO ES TRIANGULO'


    if a == b and b == c:
        return 'EQUILATERO'
    elif a == b or a == c or b == c:
        return 'ISOSCELES'
    else:
        return 'ESCALENO'
```

### Reto 05: Clasificador Cuadrático por Signo del Discriminante
**Contexto:** Para la ecuación $ax^2 + bx + c = 0$, el discriminante es $\Delta = b^2 - 4ac$. Si $a = 0$, degenera en ecuación lineal (`'LINEAL'`). Si $a \neq 0$ y $\Delta > 0$, tiene `'DOS_REALES'`. Si $\Delta == 0$, tiene `'RAIZ_DOBLE'`. Si $\Delta < 0$, tiene `'COMPLEJAS'`.

**Requerimiento:** Retorna la cadena de clasificación correspondiente.

**Firma requerida:** `reto_05(a: float, b: float, c: float)`

```python
def reto_05(a: float, b: float, c: float):
    if a == 0:
        return 'LINEAL'

    delta = (b ** 2) - (4 * a * c)

    if delta > 0:
        return 'DOS_REALES'
    elif delta == 0:
        return 'RAIZ_DOBLE'
    else:
        return 'COMPLEJAS'
```

```python
# === PRUEBAS UNITARIAS AUTOMATIZADAS (RETO 05) ===
assert reto_05(0, 4, 8) == 'LINEAL'
assert reto_05(1, -5, 6) == 'DOS_REALES'
assert reto_05(1, -2, 1) == 'RAIZ_DOBLE'
assert reto_05(1, 1, 1) == 'COMPLEJAS'
print('✅ Reto 05 Superado con éxito!')
a = float(input("Coeficiente a: "))
b = float(input("Coeficiente b: "))
c = float(input("Coeficiente c: "))
print("Clasificación:", reto_05(a, b, c))
```

### Reto 06: Suma Filtrada con Múltiplos Exclusivos
**Contexto:** Problema clásico inspirado en Project Euler. Dados los enteros positivos $N, m_1, m_2$, calcular la suma de todos los enteros positivos menores a $N$ que sean divisibles por $m_1$ o por $m_2$, pero **no** por ambos a la vez (diferencia simétrica de múltiplos).

**Requerimiento:** Retorna la suma acumulada usando un ciclo `for` y operadores lógicos.

**Firma requerida:** `reto_06(N: int, m1: int, m2: int)`

```python
def reto_06(N: int, m1: int, m2: int):
    suma = 0
    for i in range(1, N):
        div_m1 = (i % m1 == 0)
        div_m2 = (i % m2 == 0)

        # Divisible por m1 o m2, pero NO por ambos al mismo tiempo
        if (div_m1 or div_m2) and not (div_m1 and div_m2):
            suma += i

    return suma
```

### Reto 07: Conteo Posicional de Dígito Específico
**Contexto:** Conectar con el Reto 1 de BairesDev. Se escriben todos los números enteros del 1 al $N$. Se desea contar cuántas veces aparece exactamente el dígito $d$ ($0 \le d \le 9$).

**Requerimiento:** Utiliza ciclos y descomposición aritmética (división entera `// 10` y módulo `% 10`) sin convertir números a `str`.

**Firma requerida:** `reto_07(N: int, d: int)`

```python
def reto_07(N: int, d: int):
    contador = 0
    for i in range(1, N + 1):
        actual = i
        while actual > 0:
            ultimo_digito = actual % 10
            if ultimo_digito == d:
                contador += 1
            actual = actual // 10

    return contador
```

### Reto 08: Calculadora de Factorial y Ceros Finales
**Contexto:** Para un entero $n$, se calcula $n!$. Además, se desea saber cuántos ceros consecutivos tiene al final sin calcular el factorial gigante por fuerza bruta, contando los factores de 5 mediante división iterativa de Legendre.

**Requerimiento:** Retorna una tupla `(factorial, ceros_finales)`.

**Firma requerida:** `reto_08(n: int)`

```python
def reto_08(n: int):

    factorial = 1
    for i in range(1, n + 1):
        factorial *= i


    ceros_finales = 0
    potencia_5 = 5
    while n // potencia_5 > 0:
        ceros_finales += n // potencia_5
        potencia_5 *= 5

    return (factorial, ceros_finales)
```

### Reto 09: Estadística Robusta de Muestra sin Librerías
**Contexto:** Dada una lista de números reales no vacía, se requiere encontrar el valor mínimo, el valor máximo, la media aritmética y la cantidad de elementos pares (considerando solo valores enteros dentro de la lista).

**Requerimiento:** Retorna un diccionario `{'min': ..., 'max': ..., 'media': ..., 'pares': ...}` recorriendo la lista en una sola pasada `for`.

**Firma requerida:** `reto_09(numeros: list)`

```python
def reto_09(numeros: list):
    minimo = numeros[0]
    maximo = numeros[0]
    suma = 0
    pares = 0

    for num in numeros:
        if num < minimo:
            minimo = num
        if num > maximo:
            maximo = num

        suma += num

        # Verificar si es un valor entero par
        if num % 1 == 0 and num % 2 == 0:
            pares += 1

    media = round(suma / len(numeros), 2)
    return {'min': minimo, 'max': maximo, 'media': media, 'pares': pares}
```

```python
# === PRUEBAS UNITARIAS AUTOMATIZADAS (RETO 09) ===
res = reto_09([10, 5, 8, 20, 3, 12])
assert res['min'] == 3 and res['max'] == 20 and res['media'] == 9.67 and res['pares'] == 4
print('✅ Reto 09 Superado con éxito!')
entrada = input("Ingresa números separados por coma (ej: 10, 5, 8, 20): ")
numeros = []
for texto in entrada.split(','):
    numeros.append(float(texto.strip()))
print("Estadísticas:", reto_09(numeros))
```

### Reto 10: Generador de Tabla de Multiplicar Matricial Filtrada
**Contexto:** Generar una matriz bidimensional (lista de listas) de tamaño $N \times N$ con la tabla de multiplicar tradicional ($i \times j$), pero reemplazando con un cero (`0`) aquellos productos que sean múltiplos de un divisor prohibido $k$.

**Requerimiento:** Retorna la matriz resultante usando ciclos `for` anidados.

**Firma requerida:** `reto_10(n: int, k: int)`

```python
def reto_10(n: int, k: int):
    matriz = []
    for i in range(1, n + 1):
        fila = []
        for j in range(1, n + 1):
            producto = i * j
            if producto % k == 0:
                fila.append(0)
            else:
                fila.append(producto)
        matriz.append(fila)

    return matriz
```

### Reto 11: Conjetura de Collatz y Altitud Máxima
**Contexto:** La conjetura de Collatz inicia en $n$. Si $n$ es par, $n = n // 2$; si es impar, $n = 3n + 1$. El ciclo se detiene al llegar a 1.

**Requerimiento:** Usa un ciclo `while` para calcular cuántos pasos toma alcanzar el 1 y cuál fue el valor más alto alcanzado durante la trayectoria. Retorna una tupla `(pasos, max_valor)`.

**Firma requerida:** `reto_11(n: int)`

```python
def reto_11(n: int):
    pasos = 0
    max_valor = n

    while n != 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1

        pasos += 1
        if n > max_valor:
            max_valor = n

    return (pasos, max_valor)
```

```python
# === PRUEBAS UNITARIAS AUTOMATIZADAS (RETO 11) ===
assert reto_11(6) == (8, 16)
assert reto_11(11) == (14, 52)
print('✅ Reto 11 Superado con éxito!')
n = int(input("Número inicial para Collatz: "))
pasos, altitud = reto_11(n)
print(f"Pasos hasta 1: {pasos} | Valor máximo alcanzado: {altitud}")
```

### Reto 12: Algoritmo de Euclides para MCD y MCM
**Contexto:** Dados dos números enteros positivos $a$ y $b$, encontrar su Máximo Común Divisor (MCD) utilizando el algoritmo de divisiones sucesivas de Euclides con ciclo `while`. Luego, calcula su Mínimo Común Múltiplo (MCM) mediante la relación $\text{MCM}(a,b) = (a \cdot b) // \text{MCD}(a,b)$.

**Requerimiento:** Retorna la tupla `(mcd, mcm)`.

**Firma requerida:** `reto_12(a: int, b: int)`

```python
def reto_12(a: int, b: int):
    x = a
    y = b

    while y != 0:
        residuo = x % y
        x = y
        y = residuo

    mcd = x
    mcm = (a * b) // mcd
    return (mcd, mcm)
```

```python
# === PRUEBAS UNITARIAS AUTOMATIZADAS (RETO 12) ===
assert reto_12(48, 18) == (6, 144)
assert reto_12(101, 10) == (1, 1010)
print('✅ Reto 12 Superado con éxito!')
a = int(input("Primer número (a): "))
b = int(input("Segundo número (b): "))
mcd, mcm = reto_12(a, b)
print(f"MCD: {mcd} | MCM: {mcm}")
```

### Reto 13: Inversión Aritmética de Enteros y Detección de Palíndromos
**Contexto:** Un número palíndromo se lee igual de izquierda a derecha que de derecha a izquierda (ej. 12321). Se debe invertir el número aritméticamente (usando módulo `% 10` y multiplicación `* 10`) sin usar manipulación de strings (`str(n)[::-1]`).

**Requerimiento:** Retorna `True` si el número es palíndromo y `False` si no lo es.

**Firma requerida:** `reto_13(n: int)`

```python
def reto_13(n: int):
    if n < 0:
        return False

    original = n
    invertido = 0

    while n > 0:
        digito = n % 10
        invertido = (invertido * 10) + digito
        n = n // 10

    return invertido == original
```

### Reto 14: Aproximación de Raíz Cuadrada por Método Babilónico
**Contexto:** El método babilónico aproxima $\sqrt{S}$ iterando $x_{k+1} = \frac{1}{2}(x_k + \frac{S}{x_k})$. La condición de parada es $|x_{k+1} - x_k| < \epsilon$.

**Requerimiento:** Con semilla $x_0 = S / 2.0$ y tolerancia $\epsilon = 10^{-6}$, calcula la raíz aproximada redondeada a 5 decimales mediante un bucle `while`.

**Firma requerida:** `reto_14(S: float, epsilon: float = 1e-6)`

```python
def reto_14(S: float, epsilon: float = 1e-6):
    if S == 0:
        return 0.0

    x = S / 2.0
    while True:
        x_siguiente = 0.5 * (x + (S / x))
        diferencia = x_siguiente - x
        if diferencia < 0:
            diferencia = -diferencia

        if diferencia < epsilon:
            return round(x_siguiente, 5)

        x = x_siguiente
```

```python
# === PRUEBAS UNITARIAS AUTOMATIZADAS (RETO 14) ===
assert reto_14(25.0) == 5.0
assert reto_14(2.0) == 1.41421
assert reto_14(100.0) == 10.0
print('✅ Reto 14 Superado con éxito!')
S = float(input("Número para calcular raíz cuadrada: "))
print("Raíz aproximada:", reto_14(S))
```

### Reto 15: Simulador de Desintegración Radiactiva y Vidas Medias
**Contexto:** Una masa radiactiva inicial $M_0$ (en gramos) se desintegra a la mitad cada período $T$ (vida media). Se desea calcular cuántos períodos completos deben transcurrir para que la masa residual sea estrictamente menor a un umbral límite $M_{lim}$.

**Requerimiento:** Retorna una tupla `(periodos_transcurridos, masa_residual)` donde la masa esté redondeada a 4 decimales.

**Firma requerida:** `reto_15(m_inicial: float, m_limite: float)`

```python
def reto_15(m_inicial: float, m_limite: float):
    periodos = 0
    masa = m_inicial

    while masa >= m_limite:
        masa = masa / 2.0
        periodos += 1

    return (periodos, round(masa, 4))
```

### Reto 16: Generador de Números Primos en Rango
**Contexto:** Un entero $p > 1$ es primo si no tiene divisores positivos salvo 1 y él mismo. Para evaluar primalidad eficientemente, basta probar divisores hasta $\lfloor\sqrt{p}\rfloor$.

**Requerimiento:** Dado un límite superior $N$, retorna una lista ordenada con todos los números primos en el rango $[2, N]$ usando bucles y banderas booleanas.

**Firma requerida:** `reto_16(N: int)`

```python
def reto_16(N: int):
    primos = []
    for p in range(2, N + 1):
        es_primo = True
        d = 2
        while d * d <= p:
            if p % d == 0:
                es_primo = False
                break
            d += 1

        if es_primo:
            primos.append(p)

    return primos
```

```python
# === PRUEBAS UNITARIAS AUTOMATIZADAS (RETO 16) ===
assert reto_16(20) == [2, 3, 5, 7, 11, 13, 17, 19]
assert reto_16(10) == [2, 3, 5, 7]
print('✅ Reto 16 Superado con éxito!')
N = int(input("Buscar primos hasta N: "))
print("Primos:", reto_16(N))
```

### Reto 17: Búsqueda Exhaustiva de Ternas Pitagóricas
**Contexto:** Una terna pitagórica es un trío de enteros positivos $(a, b, c)$ tales que $a < b < c$ y $a^2 + b^2 = c^2$. Se busca encontrar la única terna pitagórica cuya suma exacta sea $a + b + c = S$.

**Requerimiento:** Retorna la tupla `(a, b, c)` correspondiente a la terna encontrada, o `None` si no existe ninguna para esa suma.

**Firma requerida:** `reto_17(S: int)`

```python
def reto_17(S: int):
    for a in range(1, (S // 3) + 1):
        for b in range(a + 1, ((S - a) // 2) + 1):
            c = S - a - b
            if b < c and (a * a + b * b == c * c):
                return (a, b, c)

    return None
```

```python
# === PRUEBAS UNITARIAS AUTOMATIZADAS (RETO 17) ===
assert reto_17(12) == (3, 4, 5)
assert reto_17(1000) == (200, 375, 425)
print('✅ Reto 17 Superado con éxito!')
S = int(input("Suma objetivo S para terna pitagórica: "))
print("Terna (a, b, c):", reto_17(S))
```

### Reto 18: Validador de Cuadrados Mágicos
**Contexto:** Una matriz cuadrada $N \times N$ es un cuadrado mágico si contiene todos los números de $1$ a $N^2$ sin repetir y la suma de cada fila, de cada columna y de ambas diagonales principales es idéntica a la constante mágica $M = \frac{N(N^2 + 1)}{2}$.

**Requerimiento:** Retorna `True` si la matriz es un cuadrado mágico válido y `False` en caso contrario.

**Firma requerida:** `reto_18(matriz: list)`

```python
def reto_18(matriz: list):
    n = len(matriz)
    if n == 0:
        return False

    constante_magica = (n * (n * n + 1)) // 2
    total_celdas = n * n
    vistos = [False] * (total_celdas + 1)


    for i in range(n):
        if len(matriz[i]) != n:
            return False
        suma_fila = 0
        for j in range(n):
            val = matriz[i][j]
            if val < 1 or val > total_celdas or vistos[val]:
                return False
            vistos[val] = True
            suma_fila += val
        if suma_fila != constante_magica:
            return False


    for j in range(n):
        suma_col = 0
        for i in range(n):
            suma_col += matriz[i][j]
        if suma_col != constante_magica:
            return False


    diag_principal = 0
    diag_secundaria = 0
    for i in range(n):
        diag_principal += matriz[i][i]
        diag_secundaria += matriz[i][n - 1 - i]

    if diag_principal != constante_magica or diag_secundaria != constante_magica:
        return False

    return True
```

```python

magico_3 = [[8, 1, 6], [3, 5, 7], [4, 9, 2]]
no_magico = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
assert reto_18(magico_3) == True
assert reto_18(no_magico) == False
print('✅ Reto 18 Superado con éxito!')
n = int(input("Dimensión N del cuadrado (ej: 3): "))
matriz = []
for i in range(n):
    entrada_fila = input(f"Fila {i + 1} (números separados por comas): ")
    fila = []
    for x in entrada_fila.split(','):
        fila.append(int(x.strip()))
    matriz.append(fila)
print("¿Es cuadrado mágico?:", reto_18(matriz))
```

### Reto 19: Detector de Números Felices
**Contexto:** Un número feliz es aquel en el que se sustituye el número por la suma de los cuadrados de sus dígitos repetidamente hasta que el resultado sea 1. Si entra en un ciclo sin fin que no incluye el 1 (ej. bucle que contiene el 4), no es feliz.

**Requerimiento:** Retorna `True` si $n$ es feliz y `False` si cae en ciclo infinito.

**Firma requerida:** `reto_19(n: int)`

```python
def reto_19(n: int):
    if n <= 0:
        return False

    visitados = []
    while n != 1 and n not in visitados:
        visitados.append(n)
        suma_cuadrados = 0
        temp = n
        while temp > 0:
            digito = temp % 10
            suma_cuadrados += digito * digito
            temp = temp // 10
        n = suma_cuadrados

    return n == 1
```

```python
# === PRUEBAS UNITARIAS AUTOMATIZADAS (RETO 19) ===
assert reto_19(19) == True
assert reto_19(7) == True
assert reto_19(4) == False
assert reto_19(2) == False
print('✅ Reto 19 Superado con éxito!')
n = int(input("Ingresa un entero positivo: "))
print("¿Es número feliz?:", reto_19(n))
```

### Reto 20: Reto Integrador: Criba y Extracción de Números Primos Gemelos
**Contexto:** Dos números primos $(p, q)$ son primos gemelos si difieren exactamente en 2 unidades ($q - p = 2$).

**Requerimiento:** Dado un límite $N$, encuentra todos los pares de primos gemelos en el intervalo $[2, N]$ y retorna una lista de tuplas `[(p1, q1), (p2, q2), ...]` usando ciclos y filtros booleanos.

**Firma requerida:** `reto_20(N: int)`

```python
def reto_20(N: int):
    if N < 5:
        return []


    es_primo = [True] * (N + 1)
    es_primo[0] = False
    es_primo[1] = False

    p = 2
    while p * p <= N:
        if es_primo[p]:
            for multiplo in range(p * p, N + 1, p):
                es_primo[multiplo] = False
        p += 1


    gemelos = []
    for i in range(2, N - 1):
        if es_primo[i] and es_primo[i + 2]:
            gemelos.append((i, i + 2))

    return gemelos
```

```python
# === PRUEBAS UNITARIAS AUTOMATIZADAS (RETO 20) ===
assert reto_20(20) == [(3, 5), (5, 7), (11, 13), (17, 19)]
assert reto_20(10) == [(3, 5), (5, 7)]
print('✅ Reto 20 MAESTRO Superado con éxito!')
N = int(input("Buscar primos gemelos hasta N: "))
print("Pares de primos gemelos:", reto_20(N))
```
